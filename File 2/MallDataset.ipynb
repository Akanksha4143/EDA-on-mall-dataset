```python
df["Genre"] = df["Genre"].replace({
    "Male": "M",
    "Female": "F"
})

df.head()
```

```python
df["Gender_Code"] = df["Genre"].map({
    "M": 0,
    "F": 1
})

df.head()
```

```python
#CHECKING DATA TYPES
df.dtypes
```

```python
df["CustomerID"] = df["CustomerID"].astype(str)
```

```python
df.dtypes
```

```python
#GROUP BY
df.groupby("Genre")["Spending Score (1-100)"].mean()
```

```python
df.groupby("Genre")["Annual Income (k$)"].mean()
```

```python
df.groupby("Genre")["Age"].agg(["mean", "min", "max"])
```

```python
df.groupby("Genre")["Annual Income (k$)"].agg(["mean", "min", "max"])
```

```python
plt.scatter(
    df["Annual Income (k$)"],
    df["Spending Score (1-100)"]
)

plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.show()
```

```python
plt.boxplot(df["Age"])
plt.title("Age Distribution")
plt.show()
```

```python
plt.boxplot(df["Annual Income (k$)"])
plt.title("Annual Income Distribution")
plt.show()
```

```python
df.groupby("Genre")["Spending Score (1-100)"].mean().plot(kind="bar")

plt.title("Average Spending Score by Gender")
plt.ylabel("Average Spending Score")
plt.show()
```
